# Automação de Sistemas e Processos com Python

## Desafio:

Todos os dias, o nosso sistema atualiza as vendas do dia anterior.
O seu trabalho diário, como analista, é enviar um e-mail para a diretoria, assim que começar a trabalhar, com o faturamento e a quantidade de produtos vendidos no dia anterior

E-mail da diretoria: destinatario@exemplo.com<br>
Local onde o sistema disponibiliza as vendas do dia anterior: https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga

Para resolver isso, vamos usar o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado


Referência do pyautogui: https://pyautogui.readthedocs.io/en/latest/quickstart.html

## Roteiro 

1- Entrar no sistema da empresa (link do drive) 

2- Navegar no sistema para encontrar a base de dados

3- Exportar a base de dados (baixar o arquivo)

4- Calcular os indicadores (faturamento e quantidade de produtos vendidos)

5- Enviar as informações por e-mail

### Vamos agora ler o arquivo baixado para pegar os indicadores

- Faturamento
- Quantidade de Produtos

```python
import pyautogui 
import pyperclip
import time

link = 'https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga'

time.sleep(0.5)
pyautogui.press('win')
pyautogui.write('chrome')
pyautogui.press('enter')
time.sleep(2)
pyautogui.write(link)
pyautogui.press('enter')
time.sleep(5)
pyautogui.click(x=2997, y=334, clicks=2)
time.sleep(2)
pyautogui.click(x=3072, y=333)
time.sleep(1)
pyautogui.click(x=3097, y=420)
time.sleep(5)
```

```python
import pandas as pd

caminho_arquivo = r'C:\caminho\para\Vendas - Dez.xlsx'
tabela_dados = pd.read_excel(caminho_arquivo)
display(tabela_dados)

faturamento = tabela_dados['Valor Final'].sum()
qtde_produto = tabela_dados['Quantidade'].sum()

print(faturamento)

print(qtde_produto)
```

### Vamos agora enviar um e-mail pelo gmail

```python
email = 'https://mail.google.com/'
my_email = 'destinatario@exemplo.com'
assunto_email = 'Relatório Diário de Vendas'
corpo_email = f'''
Prezados,

Encaminho o relatório de vendas referente ao dia de hoje.

Conforme os dados apresentados na planilha, seguem os resultados consolidados:

Faturamento total: R$ {f"{faturamento:,.2f}".replace(",", "X").replace(".", ",").replace("X", ".")}

Quantidade total de produtos vendidos: {f"{qtde_produto:,.0f}".replace(",", ".")} unidades.

Permaneço à disposição para quaisquer dúvidas ou esclarecimentos adicionais.

Atenciosamente,
'''

time.sleep(0.5)
pyautogui.hotkey('ctrl', 't')
time.sleep(2)
pyautogui.write(email)
pyautogui.press('enter')
time.sleep(5)
pyautogui.click(x=2652, y=179) #Clica no botão escrever do e-mail
time.sleep(3)
pyautogui.write(my_email)
pyautogui.press('tab')
pyautogui.press('tab')
time.sleep(1)
pyperclip.copy(assunto_email)
pyautogui.hotkey('ctrl', 'v')
pyautogui.press('tab')
time.sleep(1)
pyperclip.copy(corpo_email)
pyautogui.hotkey('ctrl', 'v')
pyautogui.hotkey('ctrl', 'enter')








```
